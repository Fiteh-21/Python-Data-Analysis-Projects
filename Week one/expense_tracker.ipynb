# Project 1: SpendWise — Personal Finance Tracker
### Data Analysis Journey: Week 1 Milestone
 
**Project Overview:** This project is a Python-based tool designed to track and analyze monthly expenses. It focuses on using core Python data structures to categorize spending, calculate totals, and assess financial health against a budget.

**Skills Demonstrated:**
* Data Modeling with **Lists of Dictionaries**
* Modular programming using **Functions**
* Logical flow with **Loops** and **Conditionals**
* Data presentation with **F-Strings**

```python
# The Dataset: A list of dictionaries representing a month of transactions
transactions = [
    {"date": "2026-04-01", "category": "Rent", "amount": 1200.00, "type": "Fixed"},
    {"date": "2026-04-03", "category": "Groceries", "amount": 155.50, "type": "Variable"},
    {"date": "2026-04-05", "category": "Internet", "amount": 60.00, "type": "Fixed"},
    {"date": "2026-04-07", "category": "Dining Out", "amount": 42.75, "type": "Variable"},
    {"date": "2026-04-10", "category": "Groceries", "amount": 88.20, "type": "Variable"},
    {"date": "2026-04-12", "category": "Streaming", "amount": 15.99, "type": "Fixed"},
    {"date": "2026-04-15", "category": "Transport", "amount": 35.00, "type": "Variable"},
    {"date": "2026-04-20", "category": "Gym", "amount": 50.00, "type": "Fixed"}
]

monthly_budget = 2000.00
print(f"Dataset loaded with {len(transactions)} transactions.")
```

## Section 1: Analysis Functions
In this section, we define reusable functions to process the transaction data.

```python
def calculate_total(data):
    """Calculates the total sum of all expenses."""
    total = 0
    for item in data:
        total += item["amount"]
    return total

def get_status(total, budget):
    """Compares spending to budget and returns a status message."""
    if total > budget:
        return "⚠️ OVER BUDGET"
    elif total > (budget * 0.9):
        return "🟡 WARNING: Near Limit"
    else:
        return "✅ WITHIN BUDGET"

def summarize_by_type(data):
    """Aggregates totals for Fixed vs Variable expenses."""
    summary = {"Fixed": 0, "Variable": 0}
    for item in data:
        summary[item["type"]] += item["amount"]
    return summary
```

## Reporting Engine

```python
def generate_spending_report(data, budget):
    total_spent = calculate_total(data)
    status = get_status(total_spent, budget)
    type_totals = summarize_by_type(data)
    
    # Formatting the Output
    print("=" * 50)
    print("           SPENDWISE FINANCIAL REPORT")
    print("=" * 50)
    print(f"Total Transactions: {len(data)}")
    print(f"Total Monthly Spent: ${total_spent:,.2f}")
    print(f"Monthly Budget:      ${budget:,.2f}")
    print(f"Financial Status:    {status}")
    print("-" * 50)
    
    print("SPENDING BREAKDOWN BY TYPE:")
    for expense_type, amount in type_totals.items():
        percentage = (amount / total_spent) * 100
        print(f"- {expense_type:8}: ${amount:<8,.2f} ({percentage:5.1f}%)")
    
    print("-" * 50)
    print("DETAILED LOG:")
    print(f"{'Date':12} | {'Category':12} | {'Amount':10}")
    for item in data:
        print(f"{item['date']:12} | {item['category']:12} | ${item['amount']:<9,.2f}")
    print("=" * 50)

# Execute the report
generate_spending_report(transactions, monthly_budget)
```
