# Cozy Bean Coffee Shop: Data Analysis with pandas
## Week 3 Fundamentals Workshop

This project explores the sales data of the **Cozy Bean Coffee Shop** across four locations. The goal is to demonstrate core `pandas` capabilities, moving from basic data structures to advanced filtering and transformation.

### Objectives:
*   Master **Series** and **DataFrames**.
*   Perform data inspection and quality checks.
*   Execute complex filtering and boolean masking.
*   Manipulate and sort data to extract business insights.

```python
import pandas as pd
import numpy as np
```

```python
# 1. Setup: Generate the dataset described in the workshop
def generate_data():
    dates = pd.date_range(start="2026-04-01", periods=30)
    stores = ['A', 'B', 'C', 'D']
    locations = {'A': 'Downtown', 'B': 'Airport', 'C': 'Mall', 'D': 'Suburb'}
    days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
    weather_opts = ['Sunny', 'Rainy', 'Cloudy']
    
    rows = []
    for d in dates:
        for s in stores:
            sold = np.random.randint(50, 250)
            rows.append({
                'date': d.strftime('%Y-%m-%d'),
                'store': s,
                'location': locations[s],
                'day_of_week': d.strftime('%a'),
                'weather': np.random.choice(weather_opts),
                'drinks_sold': sold,
                'revenue': sold * 3.5 if d.strftime('%a') not in ['Sat', 'Sun'] else sold * 4.0
            })
    pd.DataFrame(rows).to_csv('cozy_bean_sales.csv', index=False)
    print("File 'cozy_bean_sales.csv' created successfully!")

generate_data()
```

## Section 1 & 2: Loading and Inspecting Data
We start by loading our CSV and performing a "health check" on the data.

```python
# Load the data
df = pd.read_csv('cozy_bean_sales.csv')

# Inspection
print("--- Data Shape ---")
print(df.shape)

print("\n--- First 5 Rows ---")
display(df.head())

print("\n--- Column Info ---")
df.info()

print("\n--- Statistical Summary ---")
display(df.describe())
```

## Section 3: Selection (loc vs iloc)
*   `.loc`: Label-based selection (Inclusive).
*   `.iloc`: Integer-position based selection (Exclusive).

```python
# Selecting a specific slice
# Get the first 5 rows of 'store' and 'revenue'
selection_loc = df.loc[0:4, ['store', 'revenue']]
print("Using .loc (Label-based):")
display(selection_loc)

# Get the same data using index positions
selection_iloc = df.iloc[0:5, [1, 6]]
print("\nUsing .iloc (Position-based):")
display(selection_iloc)
```

## Section 4: Advanced Filtering
We use **Boolean Masks** to find specific performance trends, such as high-revenue days at the Mall during Sunny weather.

```python
# 1. Simple Mask: High volume days
high_volume = df[df['drinks_sold'] > 200]

# 2. Complex Mask: Mall store on Sunny days
mall_sunny = df[(df['store'] == 'C') & (df['weather'] == 'Sunny')]

# 3. Weekend Filter
weekend_sales = df[df['day_of_week'].isin(['Sat', 'Sun'])]

print(f"High Volume Days count: {len(high_volume)}")
print(f"Sunny Mall Days count: {len(mall_sunny)}")
display(mall_sunny.head())
```

## Section 5: Sorting and Business Insights
Who is our top performer? We sort by revenue to find out.

```python
# Sort by revenue descending
top_performers = df.sort_values(by='revenue', ascending=False)

print("Top 5 Highest Revenue Days:")
display(top_performers.head())
```
