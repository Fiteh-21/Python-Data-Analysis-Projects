# Project 2: Finance Pulse — Vectorized Spending Analysis
### Data Analysis Journey: Week 2 Milestone

**Project Overview:** This project elevates financial tracking by using **NumPy** to analyze a 30-day spending cycle. It transforms raw data into a structured matrix to identify weekly trends, calculate daily averages, and programmatically flag spending anomalies using high-performance vectorization.

**Skills Demonstrated:**
* **Dimensional Reshaping:** Converting flat data into 4x7 (Week x Day) matrices.
* **Axis-Wise Reductions:** Calculating performance metrics across rows and columns.
* **Vectorized Broadcasting:** Applying universal scaling (e.g., inflation/tax) without loops.
* **Boolean Masking:** Filtering "Extreme Spend" days using logical indexing.

```python
import numpy as np

def run_monthly_pulse():

    # Creating 30 days of spending data
    np.random.seed(7) 
    daily_spend = np.random.normal(loc=60, scale=25, size=30).clip(10, 200)
     
    # To make it a perfect matrix, we'll take the first 28 days for a 4x7 (Weeks x Days)
    calendar_view = daily_spend[:28].reshape(4, 7)
    print(f"Calendar View (4 Weeks x 7 Days): \n{calendar_view}\n")
    
    # Calculate the average for each day of the week
    day_averages = calendar_view.mean(axis=0)
    
    # Find "High Spend" Days 
    high_spend_mask = daily_spend > 100
    high_spend_values = daily_spend[high_spend_mask]
    high_spend_indices = np.where(high_spend_mask)[0] + 1 # +1 for 1-based calendar date
    
    
    # Aggregations & Highlights
    total_month = daily_spend.sum()
    peak_day_idx = daily_spend.argmax()
    
    # ---- Professional Report ----
    print("="*50)
    print(f"      MONTHLY FINANCE PULSE (30-DAY SUMMARY)")
    print("="*50)
    print(f"Total Monthly Expenditure:  ${total_month:,.2f}")
    print(f"Daily Average:              ${daily_spend.mean():.2f}")
    print(f"Peak Spending Date:         Day {peak_day_idx + 1} (${daily_spend.max():.2f})")
    print("-" * 50)
    
    print("WEEKDAY VS. WEEKEND ANALYSIS (Based on 4-Week Cycle):")
    days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

    for d, avg in zip(days, day_averages):
        print(f"   {d} Average: ${avg:.2f}")
    
    print("-" * 50)
    print(f"DETECTED ANOMALIES ({len(high_spend_values)} days > $100):")
    for date, val in zip(high_spend_indices, high_spend_values):
        print(f"  - Day {date}: ${val:.2f}")
    print("="*50)

run_monthly_pulse()
```
